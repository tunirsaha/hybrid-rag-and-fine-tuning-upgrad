# Run all notebooks in one Colab runtime
Runs the 7 notebooks one after another on Colab. Everything lives in Google Drive, so a runtime restart loses nothing.

1. In Google Drive create the folder **`MyDrive/hybrid_rag/`** and put `Dataset.zip` + the 7 assignment notebooks in it
2. Open this notebook in Colab -> Runtime -> Change runtime type -> **T4 GPU**
3. Run the setup cell, then the notebook cells **one at a time, top to bottom**
4. Executed copies (with outputs) are saved to `MyDrive/hybrid_rag/executed/`

**If the runtime restarts:** run the setup cell again, then continue from the notebook that was interrupted. Earlier results are already on Drive.


```python
import os, subprocess
from google.colab import drive
drive.mount('/content/drive')

WORK_DIR = '/content/drive/MyDrive/hybrid_rag'
OUT_DIR = f'{WORK_DIR}/executed'
NOTEBOOKS = [
    "Data_Understanding_and_EDA.ipynb",
    "Data_Preparation.ipynb",
    "Baseline_Model_Evaluation.ipynb",
    "RAG_Implementation.ipynb",
    "Solution_V1_RAG_Evaluation.ipynb",
    "Fine_Tuning_Pipeline.ipynb",
    "Solution_V2_FineTuned_RAG_Evaluation.ipynb"
]
missing = [n for n in NOTEBOOKS + ['Dataset.zip'] if not os.path.exists(f'{WORK_DIR}/{n}')]
assert not missing, f'Put these in MyDrive/hybrid_rag/ first: {missing}'
os.makedirs(OUT_DIR, exist_ok=True)

def run(nb):
    print(f'Running {nb} ... (no live output; this cell finishes when the notebook does)')
    r = subprocess.run(['jupyter', 'nbconvert', '--to', 'notebook', '--execute', f'{WORK_DIR}/{nb}',
                        '--output-dir', OUT_DIR, '--ExecutePreprocessor.timeout=-1',
                        '--ExecutePreprocessor.kernel_name=python3'], capture_output=True, text=True)
    if r.returncode:
        print(r.stderr[-4000:])
        raise RuntimeError(f'{nb} failed - see the error above')
    print(f'Done -> {OUT_DIR}/{nb}')

print('Ready. GPU:', subprocess.run(['nvidia-smi', '-L'], capture_output=True, text=True).stdout.strip() or 'NONE - switch to T4')
print('Already finished:', sorted(os.listdir(OUT_DIR)) or 'nothing yet')
```
